```python
df = df.fillna(df.median(numeric_only=True))

# Handle missing invoice_date (object type) by converting to datetime and filling with mode
df['invoice_date'] = pd.to_datetime(df['invoice_date'], errors='coerce')
if df['invoice_date'].isnull().any():
    most_common_date = df['invoice_date'].mode()[0]
    df['invoice_date'] = df['invoice_date'].fillna(most_common_date)

# Handle missing counter_type (object type) by filling with mode
if df['counter_type'].isnull().any():
    most_common_counter_type = df['counter_type'].mode()[0]
    df['counter_type'] = df['counter_type'].fillna(most_common_counter_type)

df.info()
```

```python
client_0_bills = df[df['client_id'] == 'train_Client_0']
print(client_0_bills)
```

```python
# Second method to select bills for client 'train_Client_0' using .query()
client_0_bills_query = df.query('client_id == "train_Client_0"')
print("\nClient 'train_Client_0' bills (Method 2 - using .query()):")
print(client_0_bills_query)
```

```python
#Transform the 'counter_type' feature to a numeric variable using the encoder of your choice.

le = LabelEncoder()
df['counter_type_encoded'] = le.fit_transform(df['counter_type'])
print(df[['counter_type', 'counter_type_encoded']].head())
```

```python
#Delete the 'counter_statue' feature from the Dataframe

df.drop('counter_statue', axis=1, inplace=True)
print(df.columns)
```
